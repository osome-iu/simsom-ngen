- **Components and code location**: 
    -`moderation` in `config/default_simulator_config.jsonc`
        - `enabled`: Toggle moderation on/off
        - `quality_threshold`: Remove messages with quality below this value (default: `0.3`)
    - In `libs/simsom/process_policy_enforcer.py`
        - Loads moderation config at startup
        - Reviews incoming messages from Workers
        - Flags messages with quality < threshold
        - Sends list of message IDs to remove to Data Manager
        - Tracks and logs moderation statistics (reviewed/flagged counts)
    - In `libs/simsom/process_data_manager.py`
        - Tracks removed message IDs in a set
        - Receives removal lists from Policy Enforcer
        - Filters removed messages from firehose and outgoing activities
        - Forwards removal IDs to Recommender via existing data channel
    - In `libs/simsom/process_data_manager.py`
        - Receives removal IDs from Data Manager
        - Filters `global_inventory` to remove flagged messages
        - Excludes removed messages when building user feeds
- **What it adds**: It adds the basic content moderation behavior (removal) based on quality.
